```python
# Load the data
def load_data(filename):
    with open(filename,"r") as f:
        data = json.load(f)
        return data
```

```python
#clean & structure the data
def clean_data(data):
    
    text_to_num = { "one" : 1 , 
                    "two" : 2,
                    "three" : 3,
                    "four" : 4,
                    "five" : 5}
    cleaned_data = []
    unique_user = set()
    
    for user in data:
        raw_rating = str(user["rating"]).strip().lower() # for remove spaces and lowercase
        if(raw_rating in text_to_num):
            raw_rating = text_to_num[raw_rating]
        # print(raw_rating)
        user["rating"] = raw_rating

        #handle missing vals
        raw_age = user.get("age")
        if(raw_age == None) : 
            user["age"] = None

        #Deduplication 
        name = user["name"].strip().lower()
        if name in unique_user:
            continue

        unique_user.add(name)
        cleaned_data.append(user)
        
    return cleaned_data

  
    
```

```python
data = clean_data(data)
```

```python
#Get meaningful insights from data
def get_insights(data):
    #avg rating
    tot_rating = 0
    for user in data:
        tot_rating += float(user["rating"])

    print(f"avg rating = {tot_rating/len(data)}")

    # % of users with poor rating
    poor_ratings = 0
    for user in data:

        if(float(user["rating"])<3):
            poor_ratings += 1

    print(f"% poor  rating = {(poor_ratings/len(data)) * 100}%")
```

```python
get_insights(data)
```

```python
#Build recommendation feature
def get_recommendation(data):
    recommendation = []
    for user in data:
        curr_recomm = {}
        curr_recomm["name"] = user["name"] 
        if(float(user["rating"]) >= 4):
            curr_recomm["brand"] = "Apple"
        else:
             curr_recomm["brand"] = "Samsung"

        recommendation.append(curr_recomm)
    return recommendation
            
        
```

```python
get_recommendation(data)
```
