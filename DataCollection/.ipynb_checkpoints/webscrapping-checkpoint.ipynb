```python
import requests
```

```python
URL = "https://www.scrapethissite.com/pages/simple/"

res = requests.get(URL)
res
```

```python
if res.status_code == 200:
    data = res.text
    print("Length:", len(data))

    f = open("data1.html", "w", encoding="utf-8")
    f.write(data)

```

```python
from bs4 import BeautifulSoup

with open("data1.html" , "r") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content , "lxml")
soup.find("h1")
```

```python
all_h3 = soup.find_all("h3")
all_countries = []
for h3 in all_h3 : 
    name = h3.get_text(strip=True)
    population = h3.find_next("div").select("span.country-population")[0].get_text()
    all_countries.append([name,population])
all_countries

import pandas as pd

df = pd.DataFrame(all_countries , columns=["name" , "population"])

df.to_csv("cleaned_data/data.csv")  
```
